```python
from pathlib import Path
import torch
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import sys

sys.path.append("../src")
```

```python
from train.train_modified_sr_corrnet import train, ModifiedSRCorrNetTrainParameters
from models.modified_sr_corrnet import (
    ModifiedSRCorrNetParameters,
    ModifiedSRCorrNetEncoderBlockParameters,
    ModifiedSRCorrNetDecoderBlockParameters,
)
```

```python
paths = sorted(list(Path("../models/modified_sr_corrnet").rglob("**/*_final.tar")))
total_runs = 1
total_losses = []
splits = []
epochs = []
run_names = []
for i, path in enumerate(paths):
    checkpoint = torch.load(path, weights_only=False)
    print(path)
    train_losses = checkpoint['train_losses']['total_loss']
    val_losses = checkpoint['val_losses']['total_loss']
    epochs_chkpt = checkpoint["train_parameters"].epochs

    total_losses += train_losses + val_losses
    splits += ["train"] * epochs_chkpt + ["val"] * epochs_chkpt
    epochs += list(range(epochs_chkpt)) + list(range(epochs_chkpt))
    run_names += [f"run-{total_runs}"] * epochs_chkpt * 2
    total_runs += 1
    # losses = pd.DataFrame({
    #     "total_losses": train_losses + val_losses,
    #     "split": ["train"] * epochs + ["val"] * epochs,
    #     "epoch": list(range(epochs)) + list(range(epochs))
    # })
    # sns.lineplot(
    #     losses,
    #     x="epoch",
    #     y="total_losses",
    #     hue="split"
    # )
    # plt.show()

losses = pd.DataFrame({
    "total_losses": total_losses,
    "split": splits,
    "epoch": epochs,
    "run_name": run_names,
})
sns.lineplot(
    losses[losses.split == "train"],
    x="epoch",
    y="total_losses",
    hue="run_name"
)
plt.show()
sns.lineplot(
    losses[losses.split == "val"],
    x="epoch",
    y="total_losses",
    hue="run_name"
)
plt.show()
    
```
