```python
import os
import torchaudio
from IPython.display import Audio, display
```

```python
DATASET_ROOT = "/media/lizi/school/705.744/Project/data/iNaturalistAudioSmall" # Set this to where your dataset is rooted
```

```python
audio, sample_rate = torchaudio.load(os.path.join(DATASET_ROOT, "mixed", "mix_0.wav"))
display(Audio(audio.numpy(), rate=sample_rate))
```

```python
import json
with open(os.path.join(DATASET_ROOT, "mixed", "mix_0.json"), "r") as f:
    metadata = json.load(f)
print(metadata)
```

```python
for source in metadata['sources']:
    source = source.replace("/media/lizi/school/705.744/Project/data/iNaturalistAudioSmall", DATASET_ROOT)
    audio, sample_rate = torchaudio.load(source)
    display(Audio(audio.numpy(), rate=sample_rate))
```
